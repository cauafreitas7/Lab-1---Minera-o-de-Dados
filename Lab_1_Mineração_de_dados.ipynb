#Calcula medidas de localização (mín., Q1, mediana, Q3, máx., média, moda), dispersão (amplitude, IQR, desvio padrão, variância, MAD), e distribuição (assimetria/skewness, curtose).

```python
import pandas as pd
import seaborn as sn
```

##Medidas de localização

```python
def minimo(df):
  return df.min()

def maximo(df):
  return df.max()

def media(df):
  return df.mean()

def mediana(df):
  return df.median()

def moda(df):
  return df.mode()

def q1(df):
  return df.quantile(0.25)

def q3(df):
  return df.quantile(0.75)
```

## Medidas de dispersão

```python
def amplitude(df):
  maximo = df.max()
  minimo = df.min()
  amplitude = maximo - minimo
  return amplitude

def iqr(df):
  q3 = df.quantile(0.75)
  q1 = df.quantile(0.25)
  distancia = q3 - q1
  return distancia

def variancia_populacional(df):
  return df.var(ddof = 0)

def desvio_padrao(df):
  return df.std(ddof = 0)

def mad(df):
  media = df.mean()
  dp_absoluto = df.sum(df - media)/df.shape[0]
```

##Medidas de dispersão




```python
def assimetria(df):
  return df.skew()

def curtose(df):
  return df.kurtosis()
```

#Detectação de outliers via regra de Tukey (1,5×IQR).

```python
titanic = sn.load_dataset("titanic")
titanic
```

```python
#Separando as variáveis em: Categóricas e Numéricas
cat_numerica = titanic[['pclass', 'age', 'sibsp', 'parch', 'fare']]
cat_categorico = titanic[['sex', 'embarked', 'class', 'who', 'adult_male', 'deck', 'embark_town', 'alive']]

def tukey(coluna):

  q1 = coluna.quantile(0.25)
  q3 = coluna.quantile(0.75)

  IQR = q3 - q1

  limite_inferior = q1 - 1.5 * IQR
  limite_superior = q3 + 1.5 * IQR

  outliers = coluna[(coluna < limite_inferior) | (coluna > limite_superior)]

  print(f"Possui {len(outliers)} outliers e a sua porcentagem é: {100*len(outliers)/len(coluna):.2f}%")

def bowley(coluna):

  q1 = coluna.quantile(0.25)
  q2 = coluna.quantile(0.50)
  q3 = coluna.quantile(0.75)

  assimetria = ((q3 -  q2) - (q2 - q1))/(q3 - q1)

  if assimetria == 0:
    return f'É assimétrica'

  elif assimetria > 0:
    return f'É assimétrica a direita(positiva)'

  elif assimetria < 0:
    return f'É assimétrica a esquerda(negativa)'

def heuristica_normalidade(coluna):

  div = abs(coluna.mean() - coluna.median())

  if div <= 0.25 * coluna.std() and abs(coluna.skew()) <= 0.5 and (-1 <= coluna.kurtosis() <= 1):
    return "Segue uma distribuiçaõ normal"

  else:
    return "Não segue uma distribuição normal"

def relatorio_numerica(dataset):
  for col in dataset:
    coluna = dataset[col].dropna()
    print(f"\n Análise da variável: {col}")
    print(f"Mínimo: {coluna.min()}")
    print(f"Q1: {coluna.quantile(0.25)}")
    print(f"Mediana: {coluna.median()}")
    print(f"Moda: {coluna.mode().iloc[0]}")
    print(f"Q3: {coluna.quantile(0.75)}")
    print(f"Máximo: {coluna.max()}")
    print(f"Média: {coluna.mean()}")
    print(f"Desvio padrão: {coluna.std()}")
    print(f"Variância: {coluna.var()}")
    print(f"MAD: {(coluna - coluna.mean()).abs().mean()}")
    print(f"Assimetria (skew): {coluna.skew()}")
    print(f"Curtose (excesso): {coluna.kurtosis()}")
    print(f"Amplitude: {coluna.max() - coluna.min()}")
    print(f"Heuristica de normalidade: {heuristica_normalidade(coluna)}")
    print(f"Assimetria: {bowley(coluna)}")
    print(f"Outliers: {tukey(coluna)}")

def relatorio_categorico(dataset):
  for coluna in dataset:
    print(f"\n Análise da variável: {coluna}")

    valores_nulos = dataset[f'{coluna}'].isnull().sum()
    print(f"Valores nulos: {valores_nulos}")

    sem_nulos = dataset[f'{coluna}'].dropna()
    frequencia = sem_nulos.value_counts()
    print(f"Frequência:\n{frequencia}")
```

```python
relatorio_numerica(cat_numerica)
```

```python
relatorio_categorico(cat_categorico)
```
